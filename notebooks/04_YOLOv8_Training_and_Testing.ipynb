```python
from ultralytics import YOLO
import glob
from IPython.display import Image, display

# 1. load the best trained model
model = YOLO('/content/runs/segment/train/weights/best.pt')

# 2.pick an image from training dataset
test_images = glob.glob(f'{dataset.location}/test/images/*.jpg')

if len(test_images) > 0:
    test_image_path = test_images[0] # taking the first image
    print(f"Testing on image: {test_image_path}")

    # 3. prediction
    results = model.predict(source=test_image_path, save=True, show_labels=False)

    # 4. show result image
    save_dir = results[0].save_dir
    image_name = test_image_path.split('/')[-1]
    display(Image(filename=f'{save_dir}/{image_name}'))
else:
    print("No images found in the test folder. Checking the validation folder...")
    val_images = glob.glob(f'{dataset.location}/valid/images/*.jpg')
    if len(val_images) > 0:
        val_image_path = val_images[0]
        # opacity hata diya hai
        results = model.predict(source=val_image_path, save=True, show_labels=False)
        save_dir = results[0].save_dir
        image_name = val_image_path.split('/')[-1]
        display(Image(filename=f'{save_dir}/{image_name}'))
    else:
        print("No test image found!")
```

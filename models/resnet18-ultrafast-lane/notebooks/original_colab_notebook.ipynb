```python
file_path = "/content/Ultrafast-Lane-Detection-Inference-Pytorch-/ultrafastLaneDetector/ultrafastLaneDetector.py"

with open(file_path, "r") as f:
    code = f.read()

# Kharab line ko dhoondh kar theek line se replace kar rahe hain
code = code.replace("return np.array(lanes_points), np.array(lanes_detected)",
                    "return np.array(lanes_points, dtype=object), np.array(lanes_detected, dtype=object)")

with open(file_path, "w") as f:
    f.write(code)

print("Library is fixed, now perform the next steps.")
```

```python
import sys
import os
import cv2
# Library path setup
sys.path.append('/content/Ultrafast-Lane-Detection-Inference-Pytorch-')
from ultrafastLaneDetector import UltrafastLaneDetector, ModelType

# 1. Paths ko fix karo
model_path = "/content/Ultra-Fast-Lane-Detection/tusimple_18.pth"
video_path = "/content/Ultra-Fast-Lane-Detection/WhatsApp Video 2026-06-30 at 14.47.04.mp4"
output_path = "/content/DL_sharp_turns.mp4" # .avi ki jagah .mp4 use karo

# 2. AI Model Load
print("Model load ho raha hai...")
lane_detector = UltrafastLaneDetector(model_path, ModelType.TUSIMPLE, use_gpu=True)

# 3. Video Capture Setup
cap = cv2.VideoCapture(video_path)
fps = int(cap.get(cv2.CAP_PROP_FPS))
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# .mp4 ke liye 'mp4v' codec sabse stable hai
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

print(f"Video processing start ho rahi hai... Output size: {width}x{height}")

# 4. Frame Processing
frame_count = 0
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # AI Lane Detection
    output_img = lane_detector.detect_lanes(frame)

    # Write frame
    out.write(output_img)

    frame_count += 1
    if frame_count % 100 == 0:
        print(f"{frame_count} frames process ho gaye...")

# 5. Finalize & Flush (Important: File corrupt hone se bachane ke liye)
cap.release()
out.release()

# Forcefully disk ko batao ki likhna khatam ho gaya
# Isse file size sahi dikhega aur play hogi
print(f"Processing done! Output yahan saved hai: {output_path}")
print(f"Final file size: {os.path.getsize(output_path) / (1024*1024):.2f} MB")
```

```python
# 1. MoviePy install karo (agar pehle se nahi hai)
!pip install moviepy

from moviepy.editor import ImageSequenceClip
import cv2
import sys
import os

# Paths
video_path = "/content/Ultra-Fast-Lane-Detection/WhatsApp Video 2026-06-30 at 14.47.04.mp4"
sys.path.append('/content/Ultrafast-Lane-Detection-Inference-Pytorch-')
from ultrafastLaneDetector import UltrafastLaneDetector, ModelType

# Model setup
model_path = "/content/Ultra-Fast-Lane-Detection/tusimple_18.pth"
lane_detector = UltrafastLaneDetector(model_path, ModelType.TUSIMPLE, use_gpu=True)

# Processing
cap = cv2.VideoCapture(video_path)
fps = int(cap.get(cv2.CAP_PROP_FPS))
frames = []

print("Frames are being processed")
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break
    # AI detection
    output_img = lane_detector.detect_lanes(frame)
    # RGB mein convert karo kyunki MoviePy RGB leta hai
    frames.append(cv2.cvtColor(output_img, cv2.COLOR_BGR2RGB))

cap.release()

# 2. MoviePy se video save karo (Corruption free)
print("Video is being saved")
clip = ImageSequenceClip(frames, fps=fps)
clip.write_videofile("/content/final_output.mp4", codec='libx264')

print("Done! download")
```
